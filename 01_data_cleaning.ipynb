```python
print(f"Original row count: {len(df):,}")

# 1. Drop rows missing ANY of our three mandatory calculation fields
df = df.dropna(subset=['Length', 'Width', 'Draft'])

print(f"Row count after removing missing physical specs: {len(df):,}")

# 2. (Optional) Quick sanity check to ensure no zero or negative dimensions slipped through
df = df[
    (df['Length'] > 0) & 
    (df['Width'] > 0) & 
    (df['Draft'] > 0)
]

print(f"Final usable row count for weight calculations: {len(df):,}")
```

```python
df_cleaned = df
```

```python
import numpy as np

# --- STEP 1: Find the baseline minimum draft for each unique ship ---
print("Calculating minimum draft baselines for each ship...")

# Group by VesselName (or MMSI) and find the absolute minimum draft in the file
# We use transform() so it automatically broadcasts the value back to matching rows without changing the row order
df_cleaned['Min_Draft_Baseline'] = df_cleaned.groupby('VesselName')['Draft'].transform('min')

# Quick sanity check: If a ship's minimum draft is unrealistically deep (e.g. over 6.5m), 
# it means we never caught it empty. We can cap the baseline at a standard empty draft of 4.5m for lakers.
df_cleaned['Min_Draft_Baseline'] = np.where(df_cleaned['Min_Draft_Baseline'] > 6.5, 4.5, df_cleaned['Min_Draft_Baseline'])


# --- STEP 2: Vectorized weight calculations (Fast & Appended as new columns) ---
print("Appending weight columns via Archimedes' Principle...")

WATER_DENSITY = 1.0  # Metric ton per cubic meter of fresh water
BLOCK_COEF = 0.88     # Hull fullness factor standard for boxy lake freighters

# Column A: Physical weight of the empty boat hull (Constant per ship)
df_cleaned['Empty_Boat_Weight_Tons'] = (
    df_cleaned['Length'] * df_cleaned['Width'] * df_cleaned['Min_Draft_Baseline'] * BLOCK_COEF * WATER_DENSITY
).round(1)

# Column B: Total current weight right now (Boat + Cargo + Fuel/Ballast)
df_cleaned['Total_Weight_Tons'] = (
    df_cleaned['Length'] * df_cleaned['Width'] * df_cleaned['Draft'] * BLOCK_COEF * WATER_DENSITY
).round(1)

# Column C: Isolated Cargo Weight (Total weight minus the empty boat weight)
# We use np.maximum to ensure that if a ship is sitting slightly below its baseline, cargo defaults to 0 instead of a negative number
df_cleaned['Cargo_Weight_Tons'] = np.maximum(0.0, df_cleaned['Total_Weight_Tons'] - df_cleaned['Empty_Boat_Weight_Tons']).round(1)

print("Weight columns successfully appended!")
print(df_cleaned[['VesselName', 'Draft', 'Min_Draft_Baseline', 'Empty_Boat_Weight_Tons', 'Cargo_Weight_Tons']].head(5))
```

```python
# 1. Define the tight, gate-to-gate lock chamber boundaries
LAT_MIN, LAT_MAX = 46.5020, 46.5050
LON_MIN, LON_MAX = -84.35600, -84.34700
46.50200, -84.356000
46.50500, -84.347000
print(f"Filtering down from {len(df_cleaned):,} regional rows...")

# 2. Extract only the pings directly around the lock gates
df_chamber = df_cleaned[
    (df_cleaned['LAT'] >= LAT_MIN) & (df_cleaned['LAT'] <= LAT_MAX) &
    (df_cleaned['LON'] >= LON_MIN) & (df_cleaned['LON'] <= LON_MAX)
].copy()

df_chamber.plot(kind='scatter',x='LON', y='LAT')

# Ensure datetimes are recognized natively by Python
df_chamber['BaseDateTime'] = pd.to_datetime(df_chamber['BaseDateTime'])

# 3. Sort chronologically so pings flow from entry to exit
df_chamber = df_chamber.sort_values(by='BaseDateTime')

# Extract the calendar date to group separate daily trips accurately
df_chamber['Date'] = df_chamber['BaseDateTime'].dt.date

print(f"Isolated {len(df_chamber):,} pings sitting directly inside the lock chambers.")

# 4. Group the continuous pings into individual transit events per ship per day
# We average the cargo weight and total weight columns across the transit row items
transit_summary = df_chamber.groupby(['VesselName', 'Length', 'Width', 'Date']).agg(
    Gate_Entry_Time=('BaseDateTime', 'min'),
    Gate_Exit_Time=('BaseDateTime', 'max'),
    Avg_Draft_Meters=('Draft', 'mean'),
    Calculated_Cargo_Tons=('Cargo_Weight_Tons', 'mean'),
    Calculated_Hull_Weight_Tons=('Empty_Boat_Weight_Tons', 'mean')
).reset_index()

# 5. Calculate total elapsed lockage time in minutes
transit_summary['Lockage_Time_Minutes'] = (
    transit_summary['Gate_Exit_Time'] - transit_summary['Gate_Entry_Time']
).dt.total_seconds() / 60.0

# 6. Apply a quick sanity check filter for a school project:
# Filter out "ghost" transits where a ship only pinged once (0 minutes)
# or weird edge cases longer than 3 hours (180 mins) due to seasonal ice lockages.
transit_summary = transit_summary[
    (transit_summary['Lockage_Time_Minutes'] > 5) & 
    (transit_summary['Lockage_Time_Minutes'] <= 180)
]

# Sort chronologically by entry time
transit_summary = transit_summary.sort_values(by='Gate_Entry_Time')

print(f"\nSuccess! Generated a clean summary of {len(transit_summary)} individual ship transits.")
```
